## Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = ["Dog bites man.", "Man bites dog.", "Dog eats meat.", "Man eats food."]
proc_docs = [doc.lower().replace(".", "") for doc in documents]
proc_docs
```

```python
vec = CountVectorizer()
bow_rep = vec.fit_transform(proc_docs)
```

```python
print("Our vocabulary: ", vec.vocabulary_)
print("BoW representation for 'dog bites man': ", bow_rep[0].toarray())
print("BoW representation for 'man bites dog': ", bow_rep[1].toarray())
```

```python
temp = vec.transform(["dog and dog are friends"])
print("BoW representation for 'dog and dog are friends': ", temp.toarray())
```

## Bag of N-Grams

```python
bon_vec = CountVectorizer(ngram_range=(1,3))
bon_rep = bon_vec.fit_transform(proc_docs)
```

```python
print("Our vocabulary: ", bon_vec.vocabulary_)
```

```python
temp = bon_vec.transform(["dog and dog are friends"])
print("BoW representation for 'dog and dog are friends': ", temp.toarray())
```

## TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer()
tfidf_rep = tfidf.fit_transform(proc_docs)
print("IDF for all words in the vocab: ", tfidf.idf_)
print("All words in the vocab: ", tfidf.get_feature_names_out())
```

```python
print("TFIDF representation for all documents: \n", tfidf_rep.toarray())
```

```python
temp = tfidf.transform(["dog and man are friends"])
print("TFIDF representation for 'dog and man are friends': \n", temp.toarray())
```

## Continuous Bag of Words (CBOW)

```python
from gensim.models import Word2Vec
```

```python
corpus = [['dog','bites','man'], ["man", "bites" ,"dog"],["dog","eats","meat"],["man", "eats","food"]]
model_cbow = Word2Vec(corpus, min_count=1, sg=0)
```

```python
print(model_cbow)
```

```python
words = list(model_cbow.wv.index_to_key)
print(words)
```

```python
print("Similarity between eats and bites:",model_cbow.wv.similarity('eats', 'bites'))
print("Similarity between eats and man:",model_cbow.wv.similarity('eats', 'man'))
```

```python
model_cbow.wv.most_similar('meat')
```

## SkipGram

```python
model_sg = Word2Vec(corpus, min_count=1, sg=1)
```

```python
print(model_sg)
```
